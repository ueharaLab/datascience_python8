# 直線回帰のためのシミュレーション


- 正規分布に従う誤差項を加えた1次関数を定義する。
- 上記関数からサンプリングを行う
- サンプリング点は以下のxを25分割した各点
\begin{align}
 0≦x≦10, a=1, b=2
\end{align}
- サンプル数は、各サンプリング点について20。

\begin{align}  
y = a + b*x + R
\end{align}
\begin{align}
R～N(0,1)
\end{align}
\begin{align}
y～N(a+bx,1)
\end{align}


```python
%matplotlib notebook
```

```python
%matplotlib inline
from numpy.random import *
import numpy as np
import matplotlib.pyplot as plt
import time
import pylab as pl
from IPython import display

x_sample = np.linspace(0,10,25)

plt.ylim(0,24)
plt.xlim(0,11)

for x in x_sample:
    for j in range(1,20):
        y = 1 + 2*x + randn()
        plt.scatter(x, y)
        display.clear_output(wait=True)
        display.display(pl.gcf())
        time.sleep(0.0001)
```
